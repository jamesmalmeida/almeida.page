# Criando Heme B Fe com Moléculas adsorvidas

### Importando as bibliotecas

```python
from ase.io import read
from ase.visualize import view
import numpy as np
from ase.build import molecule
```

### Visualizando a Heme

```python
heme_otimizada = read('optimized_heme.xyz')
view(heme_otimizada, viewer='x3d') #Visualizando
```

### Inserindo o adsorbato

```python
import copy
# Carrega a estrutura molecular relaxada da heme a partir do arquivo xyz
heme_otimizada = read('optimized_heme.xyz')

# Localiza o índice do átomo cujo número atômico é 26 (Ferro)
indice_do_ferro = np.where(heme_otimizada.numbers == 26)[0][0]
print('O indice do Fe é', indice_do_ferro)

# Extrai as coordenadas exatas (X, Y, Z) do átomo de Ferro encontrado
coordenadas_do_ferro = heme_otimizada.positions[indice_do_ferro]
print('As coordenadas do Fe são', coordenadas_do_ferro)

# Cria uma cópia independente das coordenadas do Ferro para manipular com segurança
posicao_de_vinculo_do_adsorbato = copy.deepcopy(coordenadas_do_ferro)

# Soma 4 Ångströms na coordenada Z (índice 2) para definir o ponto de ancoragem acima do Ferro
posicao_de_vinculo_do_adsorbato[2] += 4
print('As coordenadas de vínculo do adsorbato são', posicao_de_vinculo_do_adsorbato)

# Gera a estrutura idealizada da molécula de Dióxido de Carbono (CO2)
adsorbato = molecule("CO2")

# Desloca a molécula de CO2 inteira para a posição de vínculo (4Å acima do Fe)
adsorbato.positions = adsorbato.positions + posicao_de_vinculo_do_adsorbato
print(adsorbato.positions)

# Fundo as duas estruturas (Heme + CO2) em um único objeto molecular
heme_adsorbato = heme_otimizada + adsorbato

# Abre o renderizador 3D para visualizar o resultado final da junção
view(heme_adsorbato, viewer='x3d')
```

```python
heme_adsorbato.write("heme-co2.xyz")
```
