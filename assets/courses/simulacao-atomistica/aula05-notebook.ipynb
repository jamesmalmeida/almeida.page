# Tutorial de ORCA com ASE

### 1.1 - Importando bibliotecas

```python
from ase.io import read
from ase.calculators.orca import ORCA
from ase.optimize.lbfgs import LBFGS
from ase.calculators.orca import OrcaProfile
from ase.visualize import view
import numpy as np
from ase.build import molecule
```

### 1.2 - Criando, visualizando e salvando uma molécula

```python
CO2 = molecule("CO2") #Criando a molécula
view(CO2, viewer='x3d') #Visualizando
```

```python
CO2.write("CO2.xyz") #Salvando no arquivo CO2.xyz
```

## 2 - Rodando o cálculo com o ORCA

```python
# Definindo o profile to orca, basicamente o comando de execução
profile = OrcaProfile(command='/opt/apps/orca/6.1.1/orca')

# Configurando o calculator.
calc = ORCA(
    profile=profile,
    label='orca_calc',
    charge=0,  # Definindo a carga da molécula
    mult=1,   # Definindo a multiplicidade da molécula
    orcasimpleinput='B3LYP def2-TZVP Opt Freq', # Parâmetros do cálculo, 
    orcablocks='''
    %maxcore 2000 \n
    %pal nprocs 8 end'''
)

#Lendo as coordenadas da molécula.
CO2=read('CO2.xyz')
#Acoplando o calculator.
CO2.calc = calc

energy = CO2.get_potential_energy()
```

```python
CO2.write('optimized_CO2.xyz')
```
