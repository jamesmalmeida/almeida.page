```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('dados.csv', delimiter=';')
print(df)
```

```python
print(df["Tempo (s)"])
```

```python
print(df["Posição (m)"])
```

```python
df.plot.scatter("Tempo (s)", "Posição (m)")
```

```python
erro_x=df["Erro Tempo (s)"]
erro_y=df["Erro Posição (m)"]
df.plot.scatter("Tempo (s)", "Posição (m)", xerr=erro_x, yerr=erro_y, s=0.001)
```

```python
!pip install lmfit
from lmfit.models import QuadraticModel
```

```python
modelo = QuadraticModel()
```

```python
parametros = modelo.make_params()
print(parametros)
```

```python
parametros.pretty_print(columns=['value', 'min', 'max', 'vary', 'expr'])
```

```python
dados_x = df["Tempo (s)"].values
dados_y = df["Posição (m)"].values

resultado_fit = modelo.fit(dados_y, parametros, x=dados_x)
print(resultado_fit.fit_report())

resultado_fit.plot()
plt.show()
```

```python
dados_x = df["Tempo (s)"].values
dados_y = df["Posição (m)"].values

resultado_fit = modelo.fit(dados_y, parametros, x=dados_x)

SIGMA = 3
erro = resultado_fit.eval_uncertainty(sigma=SIGMA)
print(erro)

#plt.plot(dados_x, dados_y, '.', label='Pontos coletados')
plt.errorbar(dados_x, dados_y, xerr=df["Erro Posição (m)"], yerr=df["Erro Tempo (s)"], fmt='.', capsize=4, label="Dados coletados")
plt.plot(dados_x, resultado_fit.best_fit, '-', label='Melhor ajuste encontrado')
plt.fill_between(
    dados_x,
    resultado_fit.best_fit - erro,
    resultado_fit.best_fit + erro,
    color="#ABABAB",
    label=rf"Banda de confiança considerando {SIGMA}$\sigma$",
)

plt.xlabel("Tempo (s)")
plt.ylabel("Posição (m)")

plt.legend()
plt.savefig("grafico.png", dpi=300)
plt.show()
```
