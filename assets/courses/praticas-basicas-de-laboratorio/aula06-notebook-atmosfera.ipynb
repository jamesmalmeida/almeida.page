```python
import os
import glob as gb
import pandas as pd
os.getcwd()
```

```python
file = f"{os.getcwd()}\\*.txt"
lista = gb.glob(file)

todos_dados = pd.DataFrame(columns=['B','R','G','dp_B','dp_R','dp_G'])
print(todos_dados)

for i in lista:
    data = pd.read_csv(i, names=['B','R','G'])
    
    media_B = data["B"].mean()
    media_R = data["R"].mean()
    media_G = data["G"].mean()
    
    dp_B = data['B'].std()
    dp_R = data['R'].std()
    dp_G = data['G'].std()
    
    #df_tmp = pd.DataFrame(data=[media_B, media_R, media_G, dp_B, dp_R, dp_G], columns=['B','R','G','dp_B','dp_R','dp_G'])
    df_tmp = pd.DataFrame(columns=['B','R','G','dp_B','dp_R','dp_G'])
    #print(media_B, media_R, media_G, 'MEDIAS')
    print('df_tmp', df_tmp)
    df_tmp['B']=media_B
    print('TESTE', df_tmp['B'])
    df_tmp['R']=media_R
    df_tmp['G']=media_G
    df_tmp['dp_B']=dp_B
    df_tmp['dp_R']=dp_R
    df_tmp['dp_G']=dp_G
    print('df_tmp', df_tmp)
    
    df_final = pd.concat([todos_dados, df_tmp], ignore_index=True)
    
    print(df_final)
```
