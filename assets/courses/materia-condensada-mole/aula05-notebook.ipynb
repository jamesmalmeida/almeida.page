```python
import numpy as np
import matplotlib.pyplot as plt

# Set parameters (adjust these as needed)
n_c = 2.08127315132391E+021  # Critical threshold value
m = 95    # Scaling factor
max_n = 8.25902044176156E+021  # Maximum n value to plot

# Create array of n values
n = np.linspace(0, max_n, 400)

# Initialize arrays for functions
n_1 = np.zeros_like(n)
n_m = np.zeros_like(n)

# Calculate functions based on conditions
mask = n > n_c  # Boolean mask for n > n_c

# For n > n_c
n_1[mask] = n_c * (n[mask] / (m * n_c)) ** (1/m)
n_m[mask] = n[mask] / m

# For n <= n_c
n_1[~mask] = n[~mask]
n_m[~mask] = 0

# Create plot
plt.figure(figsize=(10, 6))
plt.plot(n, n_1, label='$n_1$', color='blue', linewidth=2)
plt.plot(n, n_m, label='$n_m$', color='red', linestyle='--', linewidth=2)
plt.axvline(n_c, color='gray', linestyle=':', label='$n_c$')

# Formatting
plt.title('Piecewise Functions Visualization', fontsize=14)
plt.xlabel('n', fontsize=12)
plt.ylabel('Function Values', fontsize=12)
plt.legend()
plt.grid(True, alpha=0.3)
plt.xlim(0, max_n)
plt.ylim(0, max(n_m.max(), n_1.max()) + 1)

# Show plot
plt.show()
```
